```python
async def async_function(test_param: str) -> str:
    print('This is async func')
    await asyncio.sleep(4)
    return f'Async Result: {test_param}'

async def main():
    coroutine_obj = async_function("Test")
    print(coroutine_obj)

    coroutine_result = await coroutine_obj # pauses execution of main until coroutine func finishes
    print(coroutine_result)
```

```python
import asyncio
asyncio.run(main())
```
